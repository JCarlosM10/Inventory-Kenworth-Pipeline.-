```python
import json
import random
from datetime import datetime, timedelta
import pandas as pd
from sqlalchemy import create_engine, text
```

```python
df = pd.read_json('raw_data/inventory_data_raw.json')
```

```python
df.head()
```

```python
# formatear fechas 
df['fecha_ultima_actualizacion'] = pd.to_datetime(df['fecha_ultima_actualizacion'])

#generacion de columna con valor de inventario tortal
df['valor_inventario_total'] = df['stock_actual'] * df['costo_unitario_usd']
```

```python
# definicion de bandera de nivel de stock

def flag_stock(row):
    if row['stock_actual'] == 0:
        return 'DESABASTO CRÍTICO'
    elif row['stock_actual'] <= row['stock_minimo']:
        return 'REORDEN REQUERIDO'
    elif row['stock_actual'] > row['stock_minimo']:
        return 'SOBRESTOCK'
    return 'NORMAL'

df['estado_inventario'] = df.apply(flag_stock, axis=1)
```

```python
# backup de datos limpios en formato csv y json


df.to_csv('backup_refacciones.csv', index=False)
df.to_json('backup_refacciones.json', orient='records')

# datos limpios para carga en base de datos desde formato csv o JSON mediante conexion SQLAlchemy
```
